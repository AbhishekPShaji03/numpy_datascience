```python
import numpy as np
X= np.arange(5)
print(X)

fig,axs = plt.subplots(1,2)


axs[0].plot(X,X**2,marker = 'o',markerfacecolor= 'red')
axs[0].set_title("Square")
axs[0].set_xlabel("Numbers")
axs[0].set_ylabel("Square of Numbers")

axs[1].plot(X,X**3,marker = 's',markerfacecolor= 'green')
axs[1].set_title("Cube")
axs[1].set_xlabel("Numbers")
axs[1].set_ylabel("Cube of Numbers")

plt.subplots_adjust(wspace =0.4,hspace=0.6)
plt.show()

```

```python
plt.style.use('dark_background')
import numpy as np

days = np.arange(7)

apple = [150,152,153,151,155,157,156]
google = [2800, 2820, 2810, 2830, 2840, 2850, 2860]
amazon = [3400, 3420, 3410, 3430, 3440, 3450, 3460]
microsoft = [299, 301, 300, 302, 303, 305, 304]

fig, axs = plt.subplots(2,2, figsize =(10,8))

axs[0,0].plot(days, apple, marker= 'o',color='red')
axs[0,0].set_title("Apple stock price")
axs[0,0].grid(True)

axs[0, 1].plot(days, google, marker='s', color='blue')
axs[0, 1].set_title("Google Stock Price")
axs[0, 1].grid(True)

axs[1, 0].plot(days, amazon, marker='^', color='green')
axs[1, 0].set_title("Amazon Stock Price")
axs[1, 0].grid(True)

axs[1, 1].plot(days, microsoft, marker='d', color='purple')
axs[1, 1].set_title("Microsoft Stock Price")
axs[1, 1].grid(True)

for ax in axs[1, :]:
    ax.set_xlabel("Day")
for ax in axs[:, 0]:
    ax.set_ylabel("Price ($)")

plt.tight_layout()
plt.show()
```

```python
data=[10,12,14,15,18,20,22,25,30,100]
```

```python
plt.boxplot(data,vert=False)
plt.title("Before Removing Outliers")
plt.show()
```

```python
Q1=np.percentile(data,25)
Q3=np.percentile(data,75)
print(Q1)
print(Q3)
IQR=Q3-Q1
print(IQR)
```

```python
lower_limit=Q1-(1.5*IQR)
upper_limit=Q3+(1.5*IQR)
print("Lower limit:",lower_limit)
print("Upper limit:",upper_limit)
```

```python
outliers=[x for x in data if x<lower_limit or x>upper_limit]
print(outliers)
```

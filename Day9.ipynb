```python
profit = [
12, 15, 18, 20, 22, 19, 25, 24, 21, 23,
27, 29, 31, 28, 26, 30, 32, 35, 33, 120
]

# Create a box plot for the monthly profit.
# Calculate Q1 and Q3.
# Calculate IQR.
# Calculate:
# Lower Bound
# Upper Bound
# Identify the profit outlier(s).



Sales = [
42, 45, 48, 50, 52, 54, 55, 57, 58, 60,
61, 62, 64, 65, 67, 68, 70, 72, 74, 76,
78, 80, 82, 85, 88, 120, 135, 150
]
```

```python
Q1_profit = np.percentile(profit,25)
Q3_profit = np.percentile(profit,75)

IQR_profit = Q3_profit - Q1_profit

lower_profit = Q1_profit - 1.5 * IQR_profit
upper_profit = Q3_profit + 1.5 * IQR_profit


outliers_profit = [x for x in profit if x < lower_profit or x > upper_profit]

print("PROFIT")
print("Q1 :", Q1_profit)
print("Q3 :", Q3_profit)
print("IQR :", IQR_profit)
print("Lower Bound :", lower_profit)
print("Upper Bound :", upper_profit)
print("Outliers :", outliers_profit)

Q1_Sales = np.percentile(Sales,25)
Q3_Sales = np.percentile(Sales,75)

IQR_Sales = Q3_Sales - Q1_Sales

lower_Sales = Q1_Sales - 1.5 * IQR_Sales
upper_Sales = Q3_Sales + 1.5 * IQR_Sales


outliers_Sales = [x for x in Sales if x < lower_Sales or x > upper_Sales]

print("Sales")
print("Q1 :", Q1_Sales)
print("Q3 :", Q3_Sales)
print("IQR :", IQR_Sales)
print("Lower Bound :", lower_Sales)
print("Upper Bound :", upper_Sales)
print("Outliers :", outliers_Sales)
```

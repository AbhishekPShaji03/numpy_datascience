4.Matplotlib – Practical: Given monthly sales [120, 150, 180, 140, 200], create a line plot with a title, X-axis label, Y-axis label, and grid.

```python
import matplotlib.pyplot as plt
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [120, 150, 180, 140, 200]

plt.plot(months, sales)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

plt.plot(x, y, linestyle='--', marker='v', label= 'Sales')

plt.xlabel("X Axis")
plt.ylabel("Y Axis")

plt.title("Basic Line Plot")

plt.legend()

plt.show()
```

```python
x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

plt.plot(x, y, linestyle='-.', marker='o', label= 'Sales' , color = 'red' , markerfacecolor = 'yellow' , markeredgecolor='blue')

plt.xlabel("X Axis")
plt.ylabel("Y Axis")

plt.title("Basic Line Plot")

plt.legend()

plt.show()
```
